# Partial semantic proposals from syntax

Automatic development proposals only. Scope is unassigned, parser errors remain possible, and no owner labels or B2 results are changed. Unknown fields remain explicit.


```python
from pathlib import Path
import json
import sys
import subprocess
import pandas as pd
from IPython.display import display, Markdown
current = Path.cwd()
ROOT = next((p for p in [current, *current.parents] if (p / "src/run_syntax_semantics_v1.py").exists()), None)
if ROOT is None:
    raise FileNotFoundError("Start from the project directory or notebooks folder.")
if not (ROOT / "data/processed/syntactic_candidates_v1.json").exists():
    raise FileNotFoundError("First generate the syntax output used by notebook 16.")
subprocess.run([sys.executable, str(ROOT / "src/run_syntax_semantics_v1.py")], check=True)
report = json.loads((ROOT / "data/processed/syntax_semantics_v1.json").read_text())
summary = json.loads((ROOT / "data/processed/syntax_semantics_summary_v1.json").read_text())
corpus = pd.read_csv(ROOT / "data/processed/model_corpus_v1.csv", keep_default_na=False).set_index("execution_number")
display(pd.Series(summary["target_prediction_counts"], name="automatic_prediction_count").to_frame())
print("Partial proposals only. Counts are not accuracy, reference labels or improvement over B2.")
```

```python
def show_semantic_proposals(execution_number):
    selected = [r for r in report["rows"] if r["execution_number"] == execution_number]
    if not selected:
        raise ValueError("Choose a document in the existing 20-document batch.")
    display(Markdown(f"### Execution {execution_number}"))
    display(Markdown(corpus.loc[execution_number, "model_text"]))
    display(pd.DataFrame([{"focus":r["focus"]["text"], "start":r["focus"]["start"],
        **r["proposed_fields"], "prediction":r["target_prediction"], "voice_evidence":r["voice_evidence"]} for r in selected]))
show_semantic_proposals(378)
```
