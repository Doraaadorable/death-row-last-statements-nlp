## Inspect retained decisions in context

All selected propositions were accepted after AI-assisted discussion. Acceptance is not independent annotation or verification of facts. The two nonselected documents remain in the scope table. The complete statements remain available; selected claims are not an exhaustive inventory. Corrections require a new version or a separately recorded amendment, never changing immutable evidence in place.


```python
columns = ["claim_id", "execution_number", "claim_source", "event_type", "event_actor",
           "polarity", "proposition_status", "owner_decision"]
display(claims[columns])

def show_confirmed_claim(claim_id):
    selected = claims.loc[claims.claim_id.eq(claim_id)]
    if len(selected) != 1:
        raise ValueError("Use one exact confirmed claim_id.")
    row = selected.iloc[0]
    print("Source:", row.source_url)
    print("Complete prepared statement:\n", row.model_text)
    print("Selected evidence:", row.evidence_span)
    print("Interpretation note:", row.review_note)
    display(selected[columns])

show_confirmed_claim("owner-added-524-occurrence-01")
```
